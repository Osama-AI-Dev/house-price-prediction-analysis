```python

# =========================
# 1. Import Libraries
# =========================

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
# Seaborn style
sns.set_theme(style="whitegrid")
```

```python
# =========================
# 2. Load Dataset
# =========================

df = pd.read_csv("train.csv")

print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
display(df.head())
```

```python
# =========================
# 3. Basic Information
# =========================

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())
```

```python
# =========================
# 4. Check Missing Values
# =========================

missing = df.isnull().sum()
missing_percent = (missing / len(df)) * 100

missing_df = pd.DataFrame({
    "Missing Values": missing,
    "Missing Percentage": missing_percent
})

missing_df = missing_df[
    missing_df["Missing Values"] > 0
].sort_values(
    by="Missing Percentage",
    ascending=False
)

print("\nMissing Values Before Cleaning:")
display(missing_df)
```

```python
# =========================
# 5. Remove Unnecessary Column
# =========================

# Id is only an identifier and does not represent a house feature
df.drop(columns=["Id"], inplace=True)
```

```python
# =========================
# 6. Handle Missing Values
# =========================

# ------------------------------------------------------------
# Columns where NaN means that the house does NOT have
# that particular feature.
# ------------------------------------------------------------

none_columns = [
    "PoolQC",
    "MiscFeature",
    "Alley",
    "Fence",
    "FireplaceQu",
    "GarageType",
    "GarageFinish",
    "GarageQual",
    "GarageCond",
    "BsmtQual",
    "BsmtCond",
    "BsmtExposure",
    "BsmtFinType1",
    "BsmtFinType2",
    "MasVnrType"
]

for col in none_columns:
    if col in df.columns:
        df[col] = df[col].fillna("None")
```

```python
# ------------------------------------------------------------
# Numerical columns where NaN means no such feature
# ------------------------------------------------------------

zero_columns = [
    "GarageYrBlt",
    "GarageCars",
    "GarageArea",
    "BsmtFinSF1",
    "BsmtFinSF2",
    "BsmtUnfSF",
    "TotalBsmtSF",
    "BsmtFullBath",
    "BsmtHalfBath",
    "MasVnrArea"
]

for col in zero_columns:
    if col in df.columns:
        df[col] = df[col].fillna(0)
```

```python
# ------------------------------------------------------------
# LotFrontage:
# Fill missing values using the median of each neighborhood.
# ------------------------------------------------------------

df["LotFrontage"] = df.groupby(
    "Neighborhood"
)["LotFrontage"].transform(
    lambda x: x.fillna(x.median())
)

# If any values are still missing, use global median
df["LotFrontage"] = df["LotFrontage"].fillna(
    df["LotFrontage"].median()
)
```

```python
# ------------------------------------------------------------
# Remaining numerical columns
# ------------------------------------------------------------

numeric_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
# ------------------------------------------------------------
# Remaining categorical columns
# ------------------------------------------------------------

categorical_columns = df.select_dtypes(
    include=["object"]
).columns

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# =========================
# 7. Verify Missing Values
# =========================

remaining_missing = df.isnull().sum().sum()

print("\nTotal Missing Values After Cleaning:",
      remaining_missing)

if remaining_missing == 0:
    print("✓ Dataset has no missing values.")
```

```python
# =========================
# 8. Basic Statistics
# =========================

print("\nSalePrice Statistics:")
display(df["SalePrice"].describe())


print("\nSalePrice Mean:",
      round(df["SalePrice"].mean(), 2))

print("SalePrice Median:",
      round(df["SalePrice"].median(), 2))

print("SalePrice Standard Deviation:",
      round(df["SalePrice"].std(), 2))

print("SalePrice Minimum:",
      df["SalePrice"].min())

print("SalePrice Maximum:",
      df["SalePrice"].max())



```

```python
# =========================
# 9. Distribution of SalePrice
# =========================

plt.figure(figsize=(10, 6))

sns.histplot(
    df["SalePrice"],
    kde=True
)

plt.title("Distribution of SalePrice")
plt.xlabel("Sale Price")
plt.ylabel("Frequency")

plt.tight_layout()
plt.savefig("overall_quality_vs_saleprice.png", dpi=300)
plt.show()
```

```python
# =========================
# 10. Correlation Analysis
# =========================

numeric_df = df.select_dtypes(
    include=["int64", "float64"]
)

correlation = numeric_df.corr()["SalePrice"].sort_values(
    ascending=False
)

print("\nTop Positive Correlations with SalePrice:")
display(correlation.head(11))


print("\nTop Negative Correlations with SalePrice:")
display(correlation.tail(10))
```

```python
# =========================
# 11. Top 3 Important Features
# =========================

top_features = correlation.drop(
    "SalePrice"
).sort_values(
    ascending=False
).head(3)

print("\nTop 3 Features Correlated with SalePrice:")
display(top_features)
```

```python
# ============================================================
# 12. Visualization 1
#     Overall Quality vs SalePrice
# ============================================================

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df,
    x="OverallQual",
    y="SalePrice"
)

plt.title("Overall Quality vs Sale Price")
plt.xlabel("Overall Quality")
plt.ylabel("Sale Price")

plt.tight_layout()
plt.savefig("overall_quality_vs_salepric.png", dpi=300)
plt.show()

```

```python
# ============================================================
# 13. Visualization 2
#     Living Area vs SalePrice
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="GrLivArea",
    y="SalePrice",
    alpha=0.6
)

sns.regplot(
    data=df,
    x="GrLivArea",
    y="SalePrice",
    scatter=False
)

plt.title("Living Area vs Sale Price")
plt.xlabel("Above Ground Living Area (sq ft)")
plt.ylabel("Sale Price")

plt.tight_layout()
plt.savefig("overall_quality_vs_salepri.png", dpi=300)
plt.show()
```

```python
# ============================================================
# 14. Visualization 3
#     Garage Cars vs SalePrice
# ============================================================

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="GarageCars",
    y="SalePrice"
)

plt.title("Garage Capacity vs Sale Price")
plt.xlabel("Number of Cars Garage Can Hold")
plt.ylabel("Sale Price")

plt.tight_layout()
plt.savefig("overall_quality_vs_salepr.png", dpi=300)
plt.show()

```

```python
# =========================
# 15. Additional EDA
# =========================

# Year Built vs SalePrice

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="YearBuilt",
    y="SalePrice",
    alpha=0.6
)

plt.title("Year Built vs Sale Price")
plt.xlabel("Year Built")
plt.ylabel("Sale Price")

plt.tight_layout()
plt.savefig("overall_quality_vs_salep.png", dpi=300)
plt.show()
```
